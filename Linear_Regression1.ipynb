```python
import numpy as np
import matplotlib.pyplot as plt

# Dataset
x = np.array([1, 2, 3, 4, 5, 6])
y = np.array([35, 43, 51, 59, 68, 76])

# Initial parameters
w = 0
b = 0

learning_rate = 0.01
epochs = 1000

loss_history = []


# Prediction
def predict(x, w, b):
    return w * x + b


# Mean Squared Error
def calculate_loss(y, y_pred):
    errors = y - y_pred
    squared_errors = errors ** 2
    return np.mean(squared_errors)


# Calculate gradients
def calculate_gradient(x, y, y_pred):
    n = len(x)

    dw = (-2) * np.sum(x * (y - y_pred)) / n
    db = (-2) * np.sum(y - y_pred) / n

    return dw, db


# Training
for epoch in range(epochs):

    # Make predictions
    y_pred = predict(x, w, b)

    # Calculate loss
    loss = calculate_loss(y, y_pred)

    # Store loss
    loss_history.append(loss)

    # Calculate gradients
    dw, db = calculate_gradient(x, y, y_pred)

    # Update parameters
    w = w - learning_rate * dw
    b = b - learning_rate * db


# Final results
print("Final w =", w)
print("Final b =", b)
print("Final loss =", loss_history[-1])


# Predict a new value
hours = 7
prediction = predict(hours, w, b)

print("Predicted mark for 7 hours =", prediction)
```

```python
# Plot actual data and regression line
y_pred = predict(x, w, b)

plt.scatter(x, y, label="Actual marks")
plt.plot(x, y_pred,color="red", label="Regression line")

plt.xlabel("Study Hours")
plt.ylabel("Exam Marks")
plt.title("Linear Regression From Scratch")

plt.legend()
plt.show()
```

```python
# Plot loss history
plt.plot(range(epochs), loss_history)

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Loss vs Epoch")

plt.grid()
plt.show()
```
